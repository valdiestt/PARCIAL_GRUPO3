# Simulación de la Red Social Instagram (Usuarios Estándar y VIP)

**Nombre del Sistema:** Simulación de la Red Social Instagram (Usuarios Estándar y VIP)

**Escenario a simular:** El presente proyecto simula el funcionamiento básico y las interacciones de una red social moderna, específicamente Instagram. El sistema representa un entorno digital donde los usuarios pueden crear cuentas, establecer relaciones sociales (seguir y ser seguidos), generar contenido y reaccionar a él (likes, comentarios y compartidos).

El sistema se desarrolla en **Python** mediante **programación orientada a objetos** y se organiza en tres clases principales: `Usuario`, `UsuarioVip` y `Publicacion`. La primera representa las cuentas estándar; la segunda incorpora los beneficios de una cuenta VIP; y la tercera modela las publicaciones y sus interacciones. `main.py` utiliza estas clases para realizar la simulación mediante un menú interactivo.


## 1. Clase `Usuario` — `Clase_usuario.py`

Representa una cuenta estándar de la red social. Su constructor `__init__(nombre_usuario, nombre_completo, correo)` inicializa los datos personales y las colecciones utilizadas durante la simulación.

**Atributos:**

| Atributo | Descripción |
|---|---|
| `red` | Nombre de la red social: Instagram. |
| `_nombre_usuario`, `_nombre_completo`, `_correo` | Datos identificadores de la cuenta. |
| `_seguidores`, `_seguidos` | Listas de objetos `Usuario` que representan las relaciones entre cuentas. |
| `_publicaciones` | Lista de publicaciones asociadas al usuario. |
| `_historias`, `_historias_vistas`, `_vistas_por` | Historias propias y registro de visualizaciones. |
| `_listas_publico`, `_fijadas` | Listas de público y publicaciones fijadas. |
| `_icono`, `_tipografia` | Personalización de la cuenta, inicialmente establecida como `"default"`. |

**Métodos utilizados:** `seguir(otro)` y `dejar_de_seguir(otro)` actualizan las relaciones de seguimiento; `publicar_historia(contenido)` y `ver_historia(otro)` manejan historias y visualizaciones; `crear_lista_publico(nombre)` y `fijar_publicacion(publicacion)` administran listas y publicaciones fijadas. Los métodos `obtener_*` consultan atributos y listas, mientras que `cantidad_seguidores()`, `cantidad_seguidos()` y `cantidad_publicaciones()` proporcionan estadísticas. `es_influencer()` comprueba si la cuenta tiene al menos **1 000 seguidores**.

**Reglas del plan estándar:** `DURACION_HISTORIA = 24`, `MAX_LISTAS_PUBLICO = 3` y `MAX_FIJADAS = 3`. Los permisos de visualización discreta, supercorazones, Spotlight y personalización están desactivados en esta clase.


## 2. Clase `UsuarioVip` — `usuario_vip.py`

`UsuarioVip` **hereda de `Usuario`**, por lo que conserva los datos y métodos de una cuenta normal y añade las funcionalidades de las cuentas VIP. Su constructor llama a `super().__init__(...)` para inicializar la parte heredada.

**Atributos propios:**

| Atributo | Descripción |
|---|---|
| `_insignia` | Distintivo de la membresía VIP. |
| `_beneficios_activos` | Indica si los beneficios exclusivos están habilitados. |
| `_historias_spotlight` | Lista de historias propias marcadas como destacadas. |
| `_supercorazones_enviados` | Registro de supercorazones enviados a historias de otras cuentas. |

**Métodos utilizados:** `puede_ver_discreto()`, `puede_supercorazon()`, `puede_spotlight()`, `puede_cambiar_icono()` y `puede_cambiar_tipografia()` consultan si la suscripción permite cada beneficio. `cambiar_icono(nuevo)` y `cambiar_tipografia(nueva)` personalizan la cuenta; `ver_historia(otro)` implementa la visualización discreta cuando los beneficios VIP están activos; `destacar_historia(historia)` y `obtener_historias_spotlight()` administran Spotlight; y `enviar_supercorazon(otro, historia)` junto con `obtener_supercorazones_enviados()` gestionan esa interacción exclusiva.

Además, `crear_lista_publico(nombre)` permite más de tres listas con una membresía VIP activa, pero rechaza los nombres vacíos o repetidos; `establecer_estado_plus(activo)` habilita o deshabilita los beneficios; `duracion_historia()` devuelve el valor correspondiente al plan, y `limite_fijadas()` junto con `fijar_publicacion(publicacion)` aplica el límite de publicaciones fijadas.

**Reglas de la cuenta VIP:** `DURACION_HISTORIA = 48` y `MAX_FIJADAS = 6`. Las 24 y 48 horas son valores definidos en las clases; el código no incorpora una eliminación automática de historias por tiempo transcurrido.


## 3. Clase `Publicacion` — `publicacion.py`

Representa el contenido publicado por una cuenta y las interacciones recibidas. El constructor `__init__(autor, contenido)` guarda la referencia al usuario autor y el contenido de la publicación.

**Atributos:**

| Atributo | Descripción |
|---|---|
| `_autor` | Objeto `Usuario` o `UsuarioVip` que creó la publicación. |
| `_contenido` | Texto de la publicación. |
| `_likes` | Lista de usuarios que dieron «me gusta». |
| `_comentarios` | Lista de comentarios registrados. |
| `_compartidos` | Contador de veces que se ha compartido. |

**Métodos utilizados:** `dar_like(usuario)` y `quitar_like(usuario)` agregan o retiran un «me gusta»; `comentar(usuario, texto)` registra un comentario; `compartir()` incrementa el contador de compartidos; y `cambiar_contenido(nuevo_contenido)` modifica el texto cuando el nuevo contenido es válido. Los métodos `cantidad_likes()`, `cantidad_comentarios()`, `cantidad_compartidos()` y `cantidad_interacciones()` calculan las estadísticas. `es_viral()` comprueba si las interacciones alcanzan **1 000** y `a_diccionario()` reúne los datos de la publicación en una estructura de tipo diccionario. Los métodos `obtener_*` permiten consultar sus atributos.


## 4. Relaciones entre las clases y aplicación de POO

El proyecto demuestra **herencia** mediante la relación `UsuarioVip(Usuario)`; **encapsulamiento** mediante atributos con prefijo `_` y métodos de consulta y modificación; y **polimorfismo** al redefinir en `UsuarioVip` métodos como `ver_historia()`, `crear_lista_publico()`, `duracion_historia()` y `fijar_publicacion()`.

Las clases también mantienen relaciones entre objetos: un usuario puede seguir a otros, almacenar publicaciones y aparecer como autor o como participante de los «me gusta» de una publicación. Se utilizan **listas** para representar estas asociaciones y sus interacciones.

Por último, `main.py` organiza el uso de las clases mediante un menú para consultar cuentas, gestionar seguidores, trabajar con publicaciones y administrar la membresía VIP. De esta forma, la lógica de los objetos permanece en sus respectivas clases y el archivo principal se encarga de conectarlas durante la simulación.
